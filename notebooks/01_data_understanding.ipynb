```python
PROJECT_ROOT = "/content/drive/MyDrive/heart_disease_project"
df = pd.read_csv(PROJECT_ROOT + "/data/raw/heart_disease_uci.csv")

df.info()
```

```python
df.describe()
```

```python
sns.countplot(x="thal", data=df)
plt.title("Heart Disease Distribution")
plt.show()
```

```python
print("Dataset Shape (rows, columns):", df.shape)
df.info()
```

```python
summary = df.describe()
summary
```

```python
missing = df.isna().sum()

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage (%)": (missing / len(df)) * 100
})

missing_df
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="num", data=df)
plt.title("Heart Disease Severity Distribution (num)")
plt.xlabel("Disease Level")
plt.ylabel("Count")

plt.tight_layout()
plt.savefig(EDA_REPORT_PATH + "/target_distribution.png", dpi=300)
plt.show()
```

```python
df["target_binary"] = df["num"].apply(lambda x: 1 if x > 0 else 0)

sns.countplot(x="target_binary", data=df)
plt.title("Binary Heart Disease Distribution")
plt.xticks([0,1], ["No Disease", "Disease"])
plt.savefig(EDA_REPORT_PATH + "/binary_target_distribution.png", dpi=300)
plt.show()
```

```python
numerical_cols = ["age", "trestbps", "chol", "thalch", "oldpeak", "ca"]

df[numerical_cols].hist(
    figsize=(12,8),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Numerical Feature Distributions", fontsize=14)
plt.tight_layout()
plt.savefig(EDA_REPORT_PATH + "/numerical_distributions.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df[numerical_cols])
plt.xticks(rotation=45)
plt.title("Outlier Detection via Boxplots")
plt.savefig(EDA_REPORT_PATH + "/boxplots.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df[numerical_cols + ["num"]].corr(),
            cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.savefig(EDA_REPORT_PATH + "/correlation_heatmap.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x="target_binary", y="thalch", data=df)
plt.title("Max Heart Rate vs Heart Disease")
plt.savefig(EDA_REPORT_PATH + "/thalch_vs_target.png", dpi=300)
plt.show()
```

```python
df.drop(columns=["target_binary"], inplace=True)

print("EDA completed. All plots saved to:")
print(EDA_REPORT_PATH)
```
