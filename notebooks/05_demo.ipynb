```python
from google.colab import drive
drive.mount('/content/drive')

import gradio as gr
import joblib
import numpy as np
```

```python
# Paths
PROJECT_ROOT = "/content/drive/MyDrive/heart_disease_project"
MODELS_PATH = PROJECT_ROOT + "/models"
model = joblib.load(MODELS_PATH + "/best_model.pkl")
scaler = joblib.load(MODELS_PATH + "/scaler.pkl")

print("Model and scaler loaded successfully.")
feature_names = [
    "age",
    "trestbps",
    "chol",
    "thalch",
    "oldpeak",
    "ca"
]
def predict_heart_disease(age, trestbps, chol, thalch, oldpeak, ca):
    input_data = np.array([[age, trestbps, chol, thalch, oldpeak, ca]])
    input_scaled = scaler.transform(input_data)

    prediction = model.predict(input_scaled)[0]
    probability = model.predict_proba(input_scaled)[0][1]

    if prediction == 1:
        return f"Heart Disease Detected (Probability: {probability:.2f})"
    else:
        return f"No Heart Disease Detected (Probability: {probability:.2f})"
inputs = [
    gr.Number(label="Age"),
    gr.Number(label="Resting Blood Pressure"),
    gr.Number(label="Cholesterol"),
    gr.Number(label="Max Heart Rate (thalch)"),
    gr.Number(label="Oldpeak"),
    gr.Number(label="Number of Major Vessels (ca)")
]

output = gr.Textbox(label="Prediction Result")

app = gr.Interface(
    fn=predict_heart_disease,
    inputs=inputs,
    outputs=output,
    title="Heart Disease Prediction System",
    description="Classical Machine Learning model trained on the UCI Heart Disease dataset."
)
```

```python
app.launch(share=True)
```
